```python
podcast_file_name = [f for f in os.listdir(".") if f.endswith(".m4a")][0]
print(f"Podcast file found: {podcast_file_name}")
print("Note: .m4a is audio - we need a transcript text version")
```

```python
import subprocess
import sys

subprocess.run([sys.executable, "-m", "pip", "install", "openai-whisper"], capture_output=True)

import whisper

model = whisper.load_model("base")
result = model.transcribe("The_Blueprint_For_Trustworthy_AI.m4a")
podcast_text = result["text"]

print(f"Podcast transcribed! Total characters: {len(podcast_text)}")
print(f"\nPreview:\n{podcast_text[:300]}")
```

```python
from langchain.text_splitter import CharacterTextSplitter

fixed_splitter = CharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100,
    separator=""
)

pdf_chunks_fixed = fixed_splitter.split_text(pdf_text)
podcast_chunks_fixed = fixed_splitter.split_text(podcast_text)

print(f"PDF chunks: {len(pdf_chunks_fixed)}")
print(f"Podcast chunks: {len(podcast_chunks_fixed)}")
print(f"\nFirst PDF chunk preview:\n{pdf_chunks_fixed[0][:200]}")
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    length_function=len,
    separators=["\n\n", "\n", ". ", " ", ""]
)

pdf_chunks_recursive = recursive_splitter.split_text(pdf_text)
podcast_chunks_recursive = recursive_splitter.split_text(podcast_text)

print(f"PDF recursive chunks: {len(pdf_chunks_recursive)}")
print(f"Podcast recursive chunks: {len(podcast_chunks_recursive)}")
```

```python
print("=" * 50)
print("CHUNKING COMPARISON RESULTS")
print("=" * 50)

print(f"\nPDF Document:")
print(f"  Fixed-Size:  {len(pdf_chunks_fixed)} chunks")
print(f"  Recursive:   {len(pdf_chunks_recursive)} chunks")
print(f"  Token-Based: {len(pdf_chunks_tokens)} chunks")

print(f"\nPodcast Transcript:")
print(f"  Fixed-Size:  {len(podcast_chunks_fixed)} chunks")
print(f"  Recursive:   {len(podcast_chunks_recursive)} chunks")
print(f"  Token-Based: {len(podcast_chunks_tokens)} chunks")
```

```python
from langchain.text_splitter import TokenTextSplitter

token_splitter = TokenTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

pdf_chunks_tokens = token_splitter.split_text(pdf_text)
podcast_chunks_tokens = token_splitter.split_text(podcast_text)

print(f"PDF token chunks: {len(pdf_chunks_tokens)}")
print(f"Podcast token chunks: {len(podcast_chunks_tokens)}")
```
